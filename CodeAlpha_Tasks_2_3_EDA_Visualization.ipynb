```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes_dataset.csv")
df.head()
```

```python
# Basic structure
print("Rows and columns:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isna().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Descriptive statistics
df.describe().T
```

## Task 2 — Exploratory Data Analysis

```python
# IQR-based potential outlier check
numeric_cols = df.select_dtypes(include=np.number).columns
q1 = df[numeric_cols].quantile(0.25)
q3 = df[numeric_cols].quantile(0.75)
iqr = q3 - q1

outlier_counts = ((df[numeric_cols] < (q1 - 1.5 * iqr)) |
                  (df[numeric_cols] > (q3 + 1.5 * iqr))).sum()

outlier_counts.sort_values(ascending=False)
```

```python
# Correlation with the target
target_corr = (
    df.corr(numeric_only=True)["DiseaseProgressionScore"]
      .drop("DiseaseProgressionScore")
      .sort_values(ascending=False)
)
target_corr
```

### EDA Findings

- The dataset contains **442 rows** and **11 input variables plus the target**.
- There are **0 missing values** and **0 duplicate rows**.
- The strongest positive linear relationship with the target is **bmi** (correlation ≈ 0.59).
- The strongest negative linear relationship is **s3** (correlation ≈ -0.39).
- IQR analysis identifies potential extreme observations in several numerical variables; these should be investigated rather than automatically removed.
- Correlation describes linear association and does not by itself establish causation.


## Task 3 — Data Visualization

```python
# 1. Target distribution
plt.figure(figsize=(9,5))
plt.hist(df["DiseaseProgressionScore"], bins=25, edgecolor="black")
plt.title("Distribution of Disease Progression Score")
plt.xlabel("Disease Progression Score")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# 2. Correlation heatmap
corr_all = df.corr(numeric_only=True)

plt.figure(figsize=(11,8))
plt.imshow(corr_all, aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(range(len(corr_all.columns)), corr_all.columns, rotation=70, ha="right")
plt.yticks(range(len(corr_all.index)), corr_all.index)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# 3. Strongest positive relationship
feature = "bmi"

plt.figure(figsize=(9,5))
plt.scatter(df[feature], df["DiseaseProgressionScore"], alpha=0.65)
plt.title(f"{feature} vs Disease Progression Score")
plt.xlabel(feature)
plt.ylabel("Disease Progression Score")
plt.tight_layout()
plt.show()
```

```python
# 4. Boxplots
selected = ["bmi", "bp", "s5", "DiseaseProgressionScore"]

plt.figure(figsize=(10,6))
plt.boxplot([df[c] for c in selected], labels=selected)
plt.title("Boxplots of Selected Variables")
plt.ylabel("Value")
plt.tight_layout()
plt.show()
```

```python
# 5. Compare average target across BMI quartiles
df["BMI_Group"] = pd.qcut(
    df["bmi"], q=4,
    labels=["Q1 (Lowest)", "Q2", "Q3", "Q4 (Highest)"]
)

group_means = df.groupby("BMI_Group", observed=False)["DiseaseProgressionScore"].mean()

plt.figure(figsize=(9,5))
group_means.plot(kind="bar")
plt.title("Average Disease Progression Score by BMI Quartile")
plt.xlabel("BMI Quartile")
plt.ylabel("Average Disease Progression Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

group_means
```
